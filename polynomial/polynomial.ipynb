```python
import numpy as np
import matplotlib.pyplot as plt

# Example Data (Quadratic Relationship)
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 7, 13, 21, 31])  # Notice this follows a quadratic pattern

```

```python
x_squared = x ** 2  # Square of each x

# Combine x and x² into a feature matrix X (without bias yet)
X = []
for xi, xi_sq in zip(x, x_squared):
    X.append([xi, xi_sq])

X = np.array(X)  # Shape becomes (5,2)
```

```python
ones = np.ones((X.shape[0], 1))  # Column of ones (5,1)

# Now stack [1, x, x²]
X_b = np.hstack((ones, X))  # Final shape: (5,3)
```

```python
# Theta = (Xᵗ X)^-1 Xᵗ y
X_transpose = X_b.T
theta = np.linalg.inv(X_transpose.dot(X_b)).dot(X_transpose).dot(y)

print(f"Theta values (b, m1, m2): {theta}")
```

```python
y_hat = []
for row in X_b:
    prediction = row.dot(theta)  # Multiply row with theta vector
    y_hat.append(prediction)

print(f"Predictions: {y_hat}")
```

```python
plt.scatter(x, y, color='blue', label='Actual y')
plt.plot(x, y_hat, color='red', label='Predicted y_hat')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.title('Burki’s Polynomial Regression (Degree 2)')
plt.show()
```
