```python
import matplotlib.pyplot as plt 
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
```

```python
mean_x = sum(x)/len(x)
mean_y = sum(y)/len(y)

print(mean_x)
print(mean_y)
```

```python
numerator = 0
denominator = 0
```

```python
for xi,yi in zip(x,y):
    numerator =+ (xi-mean_x)*(yi - mean_y)
    denominator =+( xi - mean_x)**2
```

```python
m = numerator/denominator
b = mean_y - m*mean_x

print(m)
print(b)
```

```python
y_hat = []

for xi in x:
    yi_hat = m*xi + b
    y_hat.append(yi_hat)
print(y_hat)    
```

```python
plt.scatter(x, y, color="blue", label="Actual y")
plt.plot(x, y_hat, color="red", label="Predicted y_hat (Line)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()  # Now it will display the labels
plt.show() 
```
