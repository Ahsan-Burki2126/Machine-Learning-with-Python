```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Data
df = pd.read_csv("../co2.csv")
x = df["Engine Size(L)"].values
y = df["CO2 Emissions(g/km)"].values
```

```python
# Hyperparameters
alpha = 0.01  # learning rate
epochs = 2010  # number of iterations
```

```python
# Initialize parameters
m = 0
b = 0
n = len(y)
```

```python
# Gradient Descent
for _ in range(epochs):
    y_hat = m * x + b
    error = y - y_hat
    dm = (-2 / n) * sum(x * error)
    db = (-2 / n) * sum(error)
    m -= alpha * dm
    b -= alpha * db
# Predictions
y_hat = m * x + b
```

```python
# MSE
mse = sum((y - y_hat) ** 2) / n
print(f"m (slope): {m}")
print(f"b (intercept): {b}")
print(f"MSE: {mse}")
```

```python
# Predict new data
new_engine = 1.5
new_prediction = m * new_engine + b
print(f"Prediction for engine size {new_engine}: {new_prediction}")
```

```python
# Plot
plt.scatter(x, y, color="blue", label="Actual Data")
plt.plot(x, y_hat, color="red", label="Predicted Line")
plt.scatter(new_engine, new_prediction, color="yellow", label="Unseen Data")
plt.xlabel("Engine Size (L)")
plt.ylabel("CO2 Emission (g/km)")
plt.legend()
plt.show()

```
