```python
import pandas as pd 
import matplotlib.pyplot as plt 
df = pd.read_csv("../co2.csv")
```

```python
x = df["Engine Size(L)"]
y = df["CO2 Emissions(g/km)"]

mean_x = sum(x)/len(x)
mean_y = sum(y)/len(y)
```

```python
numerator = 0 
denominator = 0 

for xi,yi in zip(x,y):
    numerator += (xi - mean_x)*(yi - mean_y)
    denominator += (xi - mean_x)**2

m = numerator/denominator 
b = mean_y - m*mean_x 

print(b)
print(m)
```

```python
y_hat = []
cost_sum = 0

for xi in x:
    yi_hat = m*xi + b
    y_hat.append(yi_hat)
```

```python
n = len(y)
# MSE = (yi - yi_hat)**2 for yi,yi_hat in zip(yi,yi_hat) /n

n = len(y)
error_sum = 0 
for actual,predicted in zip(y,y_hat):
    error = actual - predicted
    squared_error = error**2
    error_sum = (error_sum + squared_error)
mse = error_sum/n
print(mse)    
```

```python
new_engine = 1.5
new_prediction = m*new_engine + b


plt.scatter(x,y, color="blue", label="Actual scatter values of x,y")
plt.plot(x,y_hat,color="red",label="predicted Y hat values (Line)")
plt.scatter(new_engine,new_prediction,color="yellow",label="Plotting unseen data")
plt.xlabel("Engine Size (L)")
plt.ylabel("Co2 Emission (g/km)")
plt.legend()
plt.show()
```
