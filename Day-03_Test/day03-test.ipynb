```python
import matplotlib.pyplot as plt
```

```python
#data 
x = [1,2,3,4,5] # hours of study
y = [50,55,65,70,75,] # marks obtained
```

```python
mean_x = sum(x)/len(x)
mean_y = sum(y)/len(y)
```

```python
numerator_slope = 0
denominator_slope = 0

for xi,yi in zip(x,y):
    numerator_slope += (xi - mean_x)*(yi -mean_y)
    denominator_slope += (xi - mean_x)**2
 
```

```python
m = numerator_slope/denominator_slope
b = mean_y - m*mean_x

print(m)
print(b)
```

```python
y_hat = []
for xi in x:
    yi_hat = m*xi + b
    y_hat.append(yi_hat)
print(y_hat)    
```

```python
plt.scatter(x,y ,color ="blue" , label="actual values of y")
plt.plot(x,y_hat ,color ="red" , label="Predicted value of Y (Line)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("student hours vs marks obtained plot")
plt.legend()
plt.show()
```

```python
# Predicting marks for a student who studied 6 hours
x_new = 144
y_new_predicted = m * x_new + b

print(f"Predicted marks for {x_new} hours of study: {y_new_predicted}")
```

```python
plt.scatter(x_new, y_new_predicted, color="green", label=f"Predicted marks for {x_new} hrs")
plt.legend()
plt.show()
```
